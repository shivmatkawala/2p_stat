##### Statistics Module:

```python
import statistics
```

```python
# Descriptive Statistics:
    # 1) Measures of Central Tendency:
        # A measure of central tendancy gives a single value
        # that represents the center of typical value of a dataset
        # The three main measures are:

            # Mean => Average => center of values (sum of values/ no. of values)

            # Median => value at center after arranging it in order

            # Mode => A value which appears maximum times is called mode 
                      # dataset can be unimodal, multimodal, bimodal

    # 2) Measures of Dispersion
            # Range:- max - min

            # Variance:- the average squared distance

            # standard deviation:- the square root of the variance

            # IQR:- Q3 -Q1

            # Covariance:- (SD / Mean) * 100%

            # Mean Absolute Deviation:- ∑|x-mean| % n
    
    # 3) Measures of Position
            # Percentile
            # Quartiles
            # Z-Score

    # 4) Measures of Shape
            # Skewness: Skewness measures how assymetric the distribution is around its mean
            # Kurtosis: Kurtosis measures how heavy the tails are , meaninghow prone the data is  to extreme values compared with normal distribution
```

```python
# dataset
marks = [45, 56, 67, 87, 54, 53, 33, 20, 70, 50, 84, 90, 69, 87]

# Mean
print(statistics.mean(marks))
print(statistics.median(marks))
print(statistics.mode(marks))
```

```python
small = min(marks)
big = max(marks)

# print(small, big)

r1 = (small+big)/2
# print(r1)

from scipy.stats import iqr
print(iqr(marks))
```

```python
data = [4, 8, 2, 9, 5]
import numpy
mean_of_data = statistics.mean(data)

abs_deviations = []
for num in data:
    dev = numpy.abs(num-mean_of_data)
    abs_deviations.append(dev)

print(f"Mean Absolute Deviation: {statistics.mean(abs_deviations)}")
```

```python
std_marks = statistics.stdev(marks)
print(std_marks)

variance_marks = statistics.variance(marks)
print(variance_marks)
```

```python
study_hours = [3, 2, 5, 7, 1, 3, 3.5, 2.2]
marks = [50, 40, 70, 90, 15, 48, 65, 45]

x = statistics.correlation(study_hours, marks)
print(round(x, 2))
```

```python
price = [20, 34, 56, 78, 90]
demand = [500, 300, 200, 100, 40]

y =statistics.correlation(price, demand)
print(round(y, 2))
```

```python
salary = [10000, 22000, 12000, 34000]
show_size = [9, 7, 8, 9]

z = statistics.correlation(salary, show_size)
print(round(z, 2))
```

```python
stdev = statistics.stdev(salary)
print(stdev)

variance = statistics.variance(salary)
print(variance)

covariance = statistics.covariance(salary, show_size)
print(covariance)
```

```python
# Skewness:
data = [2, 4, 4, 6, 9]

# [∑(x-mean)**3 / n] / sd**3
print(statistics.stdev(data)**3)

```
