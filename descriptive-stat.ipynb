##### Statistics Module:

```python
import statistics
```

```python
# Descriptive Statistics:
    # 1) Measures of Central Tendency:
        # A measure of central tendancy gives a single value
        # that represents the center of typical value of a dataset
        # The three main measures are:

            # Mean => Average => center of values (sum of values/ no. of values)

            # Median => value at center after arranging it in order

            # Mode => A value which appears maximum times is called mode 
                      # dataset can be unimodal, multimodal, bimodal

    # 2) Measures of Dispersion
            # Range
            # Variance
            # standard deviation
    
    # 3) Measures of Position
            # Percentile
            # Quartiles
            # Z-Score

    # 4) Measures of Shape
            # Skewness
            # Kurtosis
```

```python
# dataset
marks = [45, 56, 67, 87, 54, 53, 33, 20, 70, 50, 84, 90, 69, 87]

# Mean
print(statistics.mean(marks))
print(statistics.median(marks))
print(statistics.mode(marks))
```

```python
small = min(marks)
big = max(marks)

# print(small, big)

r1 = (small+big)/2
# print(r1)

from scipy.stats import iqr
print(iqr(marks))
```

```python
data = [4, 8, 2, 9, 5]
import numpy
mean_of_data = statistics.mean(data)

abs_deviations = []
for num in data:
    dev = numpy.abs(num-mean_of_data)
    abs_deviations.append(dev)

print(f"Mean Absolute Deviation: {statistics.mean(abs_deviations)}")
```
